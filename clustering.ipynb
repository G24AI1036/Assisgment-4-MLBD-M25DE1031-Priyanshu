```python
def euclidean_distance(a, b):
    return np.sum((a - b) ** 2)
```

```python
def kcenter(P, k):
    centers = []

    first_center = random.choice(P)
    centers.append(first_center)

    for _ in range(1, k):
        distances = []

        for point in P:
            min_dist = min(euclidean_distance(point, center) for center in centers)
            distances.append(min_dist)

        farthest_point = P[np.argmax(distances)]
        centers.append(farthest_point)

    return np.array(centers)
```

```python
def kmeansPP(P, k):
    centers = []

    first_center = random.choice(P)
    centers.append(first_center)

    for _ in range(1, k):
        distances = []

        for point in P:
            min_dist = min(euclidean_distance(point, center) for center in centers)
            distances.append(min_dist)

        distances = np.array(distances)
        probabilities = distances / distances.sum()
        cumulative_probabilities = probabilities.cumsum()
        r = random.random()

        for idx, prob in enumerate(cumulative_probabilities):
            if r < prob:
                centers.append(P[idx])
                break

    return np.array(centers)
```

```python
def kmeansObj(P, C):
    total_distance = 0

    for point in P:
        min_dist = min(euclidean_distance(point, center) for center in C)
        total_distance += min_dist

    return total_distance / len(P)
```

```python
k = 10
k1 = 20

start = time.time()
centers_kcenter = kcenter(P, k)
end = time.time()

print('kcenter running time:', end - start)
```

```python
centers_kmeans = kmeansPP(P, k)
obj1 = kmeansObj(P, centers_kmeans)

print('KMeans++ Objective Function:', obj1)
```

```python
X = kcenter(P, k1)
C = kmeansPP(X, k)
obj2 = kmeansObj(P, C)

print('Objective Function using coreset approach:', obj2)
```
