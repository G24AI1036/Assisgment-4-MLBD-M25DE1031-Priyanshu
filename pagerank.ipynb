```python
import pandas as pd
import numpy as np
from collections import defaultdict
```

```python
file_path = 'data/small.txt'
edges = pd.read_csv(file_path, sep='\t', header=None)
edges.columns = ['source', 'destination']

edges.head()
```

```python
edges = edges.drop_duplicates()

nodes = sorted(set(edges['source']).union(set(edges['destination'])))
N = len(nodes)

print('Number of nodes:', N)
```

```python
outlinks = defaultdict(list)

for _, row in edges.iterrows():
    outlinks[row['source']].append(row['destination'])
```

```python
beta = 0.8
iterations = 40

pagerank = {node: 1 / N for node in nodes}
```

```python
for _ in range(iterations):
    new_rank = {node: (1 - beta) / N for node in nodes}

    for node in nodes:
        outgoing = outlinks[node]

        if len(outgoing) > 0:
            share = beta * pagerank[node] / len(outgoing)

            for dest in outgoing:
                new_rank[dest] += share

    pagerank = new_rank
```

```python
top_5 = sorted(pagerank.items(), key=lambda x: x[1], reverse=True)[:5]
bottom_5 = sorted(pagerank.items(), key=lambda x: x[1])[:5]

print('Top 5 Nodes:')
for node, score in top_5:
    print(node, score)

print('\nBottom 5 Nodes:')
for node, score in bottom_5:
    print(node, score)
```
